```python
import ipywidgets as widgets
from IPython.display import display

# Create the slider widget
slider = widgets.IntSlider(
    value=50,
    min=0,
    max=100,
    step=1,
    description='Value:'
)

# Create the output widget to display the text
output = widgets.Output()

# Define the callback function for when the slider moves
def on_value_change(change):
    output.clear_output()
    with output:
        print(f'Slider value is {change["new"]}')

# Monitor the slider's value for changes
slider.observe(on_value_change, names='value')

# Display both widgets in the notebook
display(slider, output)
```
